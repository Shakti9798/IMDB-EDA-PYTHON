### Q3.Clean the 'duration' column . (e.g., '175min' - 175)

```python
df["duration"] = df["duration"].str.replace("min","")
```

```python
df["duration"] = df["duration"].str.replace(",","")
```

```python
df["duration"] = df["duration"].astype(int)
```

```python
df.info()
```

```python
df["duration"].mean()
```

### Q4.Clean the 'votes' and 'gross_income' column.

```python
df["votes"] =  df["votes"].str.replace(",","")
```

```python
df["votes"] = df["votes"].astype(int)
```

```python
df.info()
```

```python
df["gross_income"] = df["gross_income"].str.replace(",","")
```

```python
df["gross_income"] = df["gross_income"].str.replace("$","")
```

```python
df["gross_income"] = df["gross_income"].str.replace("M","")
```

```python
df["gross_income"] = df["gross_income"].astype(float)
```

# **VISUALIZATION**

```python
import matplotlib.pyplot as plt
```

### Q1. Average rating over years.

```python
roy = df.groupby("year")["rating"].mean().sort_index()
```

```python
plt.figure(figsize = (15,6))

plt.plot(roy.index, roy.values, color = "darkred", marker = "d", markersize = 5, linestyle = "-.")

plt.title("Average Rating Over The Years")
plt.xlabel("Years")
plt.ylabel("Average Rating")

plt.grid(alpha = 0.3)

plt.show()
```

### Q2. Number of movies released per year.

```python
plt.figure(figsize = (15,6))

mpy = df.groupby("year")["id"].count()

plt.plot(mpy.index, mpy.values, color = "blue", marker = "*", markersize = 4, linestyle = "dotted")
plt.title("Movies Released Per Year")
plt.xlabel("Years")
plt.ylabel("Number of Movies")

plt.grid(alpha = 0.3)

plt.show()
```

### Q3. Rating distribution.

```python
plt.figure(figsize = (15,6))

plt.hist(df["rating"], color = "gray", edgecolor = "black", bins = 30)

plt.title("Distribution Of Movie Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")

plt.show()
```

### Q4. Which are the top 10 most frequent actors? (Horizontal Bar Chart)

```python
x = df["stars_name"].str.split(",").explode()
```

```python
top_10 = x.value_counts().sort_values(ascending = False).head(10)
```

```python
top_10
```

```python
plt.figure(figsize = (15,6))

plt.barh(top_10.index, top_10.values, color = ["skyblue", "blue", "darkblue","red", "violet"])

plt.title("Most Frequent Stars")
plt.xlabel("stars_name")
plt.ylabel("Frequency")

plt.xticks(rotation = 45)

plt.show()
```

### Q5. Relationship between year and rating (Scatter Plot)

### Q8.Count of movies for the top actors (Count Plot using Seaborn)

```python
a = df["stars_name"].str.split(",").explode().reset_index(drop=True)

top_actors = a.value_counts().head(10).index

plt.figure(figsize=(10,6))

sns.countplot(y = a, order = top_actors, hue = a, palette = "coolwarm", legend = False)
plt.title("Top Actor Appearances")
plt.xlabel("Number of Appearances")
plt.ylabel("Actor")

plt.show()
```
